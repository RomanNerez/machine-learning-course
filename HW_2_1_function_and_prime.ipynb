```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
```

```python
xs = np.linspace(0, 20, 100)

def f(x):
    return np.sin(x)

def f_prime(x):
    return np.cos(x)

x0 = xs[39]
x1 = xs[20]

y0 = f(x0)
y1 = f(x1)
k0 = f_prime(x0)
k1 = f_prime(x1)

y_tangent0 = y0 + k0 * (xs - x0)
y_tangent1 = y1 + k1 * (xs - x1)

plt.plot(xs, f(xs), label="Функція f(x) = x^2")
plt.plot(xs, y_tangent0, 'r', label=f"Дотична в x0 = {x0:.2f}")
plt.plot(xs, y_tangent1, 'r', label=f"Дотична в x1 = {x1:.2f}")
plt.scatter([x0], [y0], color="black", zorder=5, label="Точка дотику (x0)")
plt.scatter([x1], [y1], color="black", zorder=5, label="Точка дотику (x1)")
plt.legend()
plt.grid(True)
plt.show()
```

1. Задана функція залежності оцінки від кількості годин, витрачених на навчання: $ f(x) = 10x - 5 $.
  1. З допомогою Python обчисліть похідну цієї функції.
  2. Використовуйте бібліотеку `matplotlib` для побудови графіків функції та її похідної в діапазоні значень від 0 до 10 годин. Проаналізуйте, як зміна кількості годин вивчення впливає на оцінку.



```python
xs = np.linspace(0, 10, 100)

def f(x):
  return 10 * x - 5

def f_prime(x):
  return np.array([10] * len(x))


plt.plot(xs, f(xs))
plt.plot(xs, f_prime(xs), 'r')
plt.grid(True)
plt.show()
```

**Висновок**: так як похідна це константа, то з кожною годиною оцінка стабільно збільшується на 10

1 година = 10 * 1 - 5 = 5

2 година = 10 * 2 - 5 = 15

3 година = 10 * 3 - 5 = 25

Різниця між годинами дорівнює 10

2. Використовуючи Python і `SymPy`, обчисліть похідні для функцій $ f(x) = \sin(x) $ та $ f(x) = \ln(x) $. Зобразіть ці функції та їхні похідні за допомогою `matplotlib`.



```python
x_sym = sp.Symbol('x')
f_sym_sin = sp.sin(x_sym)
f_prime_sym_sin = sp.diff(f_sym_sin, x_sym)

print(f"Початкова функція: f(x) = {f_sym_sin}")
print(f"Похідна функції: f'(x) = {f_prime_sym_sin}")

f_sym_ln = sp.ln(x_sym)
f_prime_sym_ln = sp.diff(f_sym_ln, x_sym)

print(f"Початкова функція: f(x) = {f_sym_ln}")
print(f"Похідна функції: f'(x) = {f_prime_sym_ln}")
```

```python
xs = np.linspace(0, 10, 100)

f_num = sp.lambdify(x_sym, f_sym_sin, 'numpy')
f_prime_num = sp.lambdify(x_sym, f_prime_sym_sin, 'numpy')

plt.plot(xs, f_num(xs))
plt.plot(xs, f_prime_num(xs), 'r')
plt.grid(True)
plt.show()
```

```python
xs = np.linspace(1, 10, 100)

f_num = sp.lambdify(x_sym, f_sym_ln, 'numpy')
f_prime_num = sp.lambdify(x_sym, f_prime_sym_ln, 'numpy')

plt.plot(xs, f_num(xs))
plt.plot(xs, f_prime_num(xs), 'r')
plt.grid(True)
plt.show()
```


3. Загальні витрати компанії від обсягу виробництва задані функцією $ f(x) = x^2 - 5x + 1 $, де $ x $ — кількість вироблених одиниць. Обчисліть похідну цієї функції з допомогою SymPy та візуалізуйте значення функції і її похідної на відрізку від 0 до 100.

```python
x_sym = sp.Symbol('x')
f_sym = x_sym ** 2 - 5 * x_sym + 1
f_prime_sym = sp.diff(f_sym_sin, x_sym)

print(f"Початкова функція: f(x) = {f_sym}")
print(f"Похідна функції: f'(x) = {f_prime_sym}")
```

```python
xs = np.linspace(0, 100, 10)

f_num = sp.lambdify(x_sym, f_sym, 'numpy')
f_prime_num = sp.lambdify(x_sym, f_prime_sym, 'numpy')

plt.plot(xs, f_num(xs))
plt.plot(xs, f_prime_num(xs), 'r')
plt.grid(True)
plt.show()
```
