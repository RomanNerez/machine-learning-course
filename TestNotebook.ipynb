### Друга дія це продовження будування колайдера з тих елементів які ми маємо

```python
def sum(x):
    return x + 5

list = [sum(x) for x in range(10)]

print(list)
```

### Третя дія це можливість розширення колайдера

```python
import glob

# 1. знаходження документації колайдера
txt_files = glob.glob("*.txt")

# 2. знаходження його фото
all_images = glob.glob("images/**/*.jpg", recursive=True)

# 3. знаходження його дані на яких він навчався
specific_data = glob.glob("data[0-9].csv")

print(txt_files, all_images, specific_data) # нажаль, нічого немає, але ми знайдемо!
```
